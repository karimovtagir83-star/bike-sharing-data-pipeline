```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("../data/hour.csv")
```

```python
peak_hour = df.groupby("hr", as_index=False)["cnt"].mean()
peak_hour.sort_values("cnt", ascending=False)

```

```python
print(df.columns)
```

```python
result = df.groupby("workingday", as_index=False)["cnt"].mean()
result = result.T
result = result.drop(index="workingday")
result = result.rename(columns={
    0: "non_working_day",
    1: "working_day"
})
result["difference"] = (result["non_working_day"] - result["working_day"])
print(result)

```

```python
result2 = df.groupby("season", as_index=False)["cnt"].mean()
result2 = result2.T
result2 = result2.drop(index="season")
result2 = result2.rename(index={
    "cnt": "mean_cnt"
})
result2 = result2.rename(columns={
    0: "winter",
    1: "spring",
    2: "summer",
    3: "autumn"
})
print(result2)
```

```python
result3 = df.groupby("dteday", as_index=False)["cnt"].sum()
result3 = result3.sort_values("cnt", ascending=False)
result3 = result3.head(5)
print(result3)
```

```python
fig = px.bar(result3, x="dteday", y="cnt")
fig.show()
```

```python
result4 = df.groupby(["season", "hr"], as_index=False)["cnt"].mean()
indexes = result4.groupby("season")["cnt"].idxmax()
print(result4.loc[indexes])
```

```python
result5 = df[df["workingday"] == 1]
result5 = result5.groupby(["season", "weekday"], as_index=False)["cnt"].mean()
indexes = result5.groupby("season")["cnt"].idxmax()
print(result5.loc[indexes])
```

```python
fig = px.line(peak_hour, x="hr", y="cnt")
fig.show()
```

```python
seasons = ["winter", "spring", "summer", "autumn"]

fig = px.bar(
    x=seasons,
    y=result2.loc["mean_cnt", seasons]
)

fig.show()
```
